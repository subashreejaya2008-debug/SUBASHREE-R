```python
#Loading Dataset
df_airline = pd.read_csv("/content/airlines_flights_data.csv")

print("Dataset loaded successfully")
```

Task 1:Single column groupby aggregation

Question:Calculate the average departure delay for each airline carrier

```python
#code
airline_col = 'airline'
dep_col = 'departure_time'
origin_col = 'source_city'
dest_col = 'destination_city'

task1_airline = df_airline.groupby(airline_col)[dep_col].count().reset_index(name='Flight_Count')

print(task1_airline.head(5))

```

Task 2:Multi column groupby aggregation
Question:Calculate total flights and the average departure delays grouped by origin and destination airports

```python
#code
task2_airline = df_airline.groupby([origin_col, dest_col]).agg(
   Flight_Count=(dep_col, 'count'),
).reset_index()
print(task2_airline.head(5))
```

Task3: Multi-MetricAggregation with agg()

Question:Compute total flight count,mean departure delay and maximum departure delay per airline



```python
#code
task3_airline = df_airline.groupby(airline_col).agg(
    Total_Flight_Count=(dep_col, 'count'),
).reset_index()
print(task3_airline.head(5))
```

Task4:Sorting Aggregation Summaries

Question:Identify the top 5 origin airports handling the highest volume of departing flights

```python
#code
task4_airline = (df_airline.groupby(origin_col)
                .size()
                .reset_index(name='Total_Flights')
                .sort_values(by='Total_Flights', ascending=False)
                .head(5))
print(task4_airline)
```

Task 5:Named Aggregations

Questions:Compute total flight count,average departure delay and average arrival delay per airline.

```python
#code
arr_col = 'arrival_time'
task5_airline = (df_airline.groupby(airline_col).agg(
                    Total_Flights=(dep_col, 'count')
).reset_index())
print(task5_airline.head(5))
```

Task 6:Conditional group filtering

Question:Filter the dataset to re-evaluate flights origination only from hub airports that handle over 10000 departing flights



```python
#code
task6_airline = df_airline.groupby(origin_col).filter(lambda x: x[dep_col].count() > 10000)
print("Qualified Origin Airports")
print(task6_airline[origin_col].unique())
print("Hub Flights Retained:", len(task6_airline))
```

Task7: Derived aggregate matrix

Question:Calculate total flights,delayed flights (DEP_DELAY > 15), and Delayed Rate Percentage Per airline



```python
#code
task7_airline =  df_airline.groupby(airline_col).agg(
    Total_Flights=(dep_col, 'count')
).reset_index()
print(task7_airline.sort_values(by='Total_Flights', ascending=False).head(5))
```

Task 8:Extremum selection with nlargest()

Question:Retrive the top 5 individual flight instances recording the worst overall departure delay.



```python
#code
task8_airline = df_airline.groupby(airline_col).size().nlargest(5).reset_index(name='Total_Flights')
print("Top 5 airlines by total flights:")
print(task8_airline)
```

Task 9:Quantile Aggregation

Question:Calculate the 90th percentile (P90)departure delay value for each airline carrier

```python
#code
task9_airline = df_airline.groupby(airline_col)[dep_col].value_counts().unstack(fill_value=0)
print("Departure Time Distribution by Airline (Counts):")
print(task9_airline)
```

Task 10: Frequency Counting via Groupby

Question:Identify the top 10 busiest Origin-to-Destination flights routes by total flight volume



```python
#code
task10_airline = (df_airline.groupby([origin_col, dest_col])
                 .size()
                 .reset_index(name='Flight_Count')
                 .sort_values(by='Flight_Count', ascending=False)
                 .head(10))
print(task10_airline)
```

```python
#code
import numpy as np

bins = [0, 5, 15, np.inf]
labels = ['Short Haul', 'Medium Haul', 'Long Haul']

df_airline['Flight_Category'] = pd.cut(df_airline['duration'], bins=bins, labels=labels, right=False)

display(df_airline.head())
```

Task 1: Plot Flight Price Distribution with KDE

Task4:Flight Duration vs Price Scatter Plot split by class

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='duration',
    y='price',
    hue='class',
    alpha=0.5
)

plt.title('Flight Duration vs Price by Travel Class')
plt.xlabel('Flight Duration')
plt.ylabel('Price')

plt.show()
```

Task 5: Heatmap Matrix of Mean Prices between Souces and Destination

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a pivot table of mean price
price_pivot = df.pivot_table(
    values='price',
    index='source_city',
    columns='destination_city',
    aggfunc='mean'
)

# Create the heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    price_pivot,
    annot=True,
    fmt='.0f',
    cmap='YlGnBu'
)

plt.title('Mean Flight Price by Source and Destination City')
plt.xlabel('Destination City')
plt.ylabel('Source City')

plt.show()
```

Task 6:Average Price by Departure Time Bar plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x='departure_time',
    y='price',
    estimator='mean'
)

plt.title('Average Flight Price Across Departure Time Slots')
plt.xlabel('Departure Time')
plt.ylabel('Average Price')
plt.xticks(rotation=45)

plt.show()
```

Task 7:Correlation Matrix of Numerical Flight Attributes

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a copy of the required columns
corr_df = df[['duration', 'price', 'stops']].copy()

# Convert stops to numeric values
stops_mapping = {
    'zero': 0,
    'one': 1,
    'two_or_more': 2
}

corr_df['stops'] = corr_df['stops'].map(stops_mapping)

# Compute correlation matrix
correlation_matrix = corr_df.corr()

print(correlation_matrix)

# Plot correlation heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix: Duration, Price, and Stops')

plt.show()
```

Task 8:Horizontal Countplot of Flights per Airline

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y='airline'
)

plt.title('Number of Available Flights per Airline')
plt.xlabel('Number of Flights')
plt.ylabel('Airline')

plt.show()
```

Task 9:Cross-Tabulate Counts of Flights between Airline and stops

```python
crosstab = pd.crosstab(df['airline'], df['stops'])

print(crosstab)
```

Task 10: Identify top 3 longest duration flights for each airline

```python
top_3_flights = df.groupby('airline').apply(
    lambda x: x.nlargest(3, 'duration')
).reset_index(drop=True)

print(top_3_flights)
```

```python
# AIRLINE  DATASET

import kagglehub

# Download latest version
path = kagglehub.dataset_download("rohitgrewal/airlines-flights-data")

print("Path to dataset files:", path)


import os

print(os.listdir(path))


import pandas as pd

df = pd.read_csv(os.path.join(path, "airlines_flights_data.csv"))

print( "Dataset loaded successfully")
```

Task1: Calculate mean flight price per airline using tranform()

```python
# Calculate the average price for each airline
# and assign the average to every corresponding row

df['Avg_Airline_Price'] = df.groupby('airline')['price'].transform('mean')

# Display the result
print(df[['airline', 'price', 'Avg_Airline_Price']].head())
```

Task2: Calculate Price Deviation from Airline Average

```python
# Calculate the difference between the actual price
# and the average price for that airline

df['Price_Difference'] = df['price'] - df['Avg_Airline_Price']

# Display the result
print(df[['airline', 'price', 'Avg_Airline_Price', 'Price_Difference']].head())
```

Task 3:Create pivot table of average price by source city and class

```python
# Create pivot table with Source City as index,
# travel class as columns, and average price as values

pivot_table = df.pivot_table(
    index='source_city',
    columns='class',
    values='price',
    aggfunc='mean'
)

print(pivot_table)
```

Task 4:Categorize Duration into flight distance categories

```python
import pandas as pd

bins = [0, 3, 6, df['duration'].max()]
labels = ['Short Haul', 'Medium Haul', 'Long Haul']

df['Duration_Category'] = pd.cut(
    df['duration'],
    bins=bins,
    labels=labels,
    include_lowest=True
)

print(df[['duration', 'Duration_Category']].head())
```

Task 5:Extract Flight Prefix Code using Vectorized String Operations

```python
df['Airline_Code'] = df['flight'].str.split('-').str[0]

print(df[['flight', 'Airline_Code']].head())
```

Task 6: Filter Flights Departing in early morning or morning with price above median

```python
median_price = df['price'].median()

filtered_df = df[
    (df['departure_time'].isin(['Early_Morning', 'Morning'])) &
    (df['price'] > median_price)
]

print(filtered_df)
```

Task 7: Create a Route Column Combining Source and Destination Cities

```python
df['Route'] = df['source_city'] + ' > ' + df['destination_city']

print(df[['source_city', 'destination_city', 'Route']].head())
```

Task 8: Calculate Ranking of Flights by Price within Each Airline

```python
df['Price_Rank'] = df.groupby('airline')['price'].rank(ascending=False)

print(df[['airline', 'price', 'Price_Rank']].head())
```

Task 9:Cross-Tabulate Counts of Flights between Airline and stops

```python
crosstab = pd.crosstab(df['airline'], df['stops'])

print(crosstab)
```

Task 10: Identify top 3 longest duration flights for each airline



```python
top_3_flights = df.groupby('airline').apply(
    lambda x: x.nlargest(3, 'duration')
).reset_index(drop=True)

print(top_3_flights)
```
